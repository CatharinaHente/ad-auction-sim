```python
import sys
sys.path.insert(0, '/Users/catharinahente/Documents/Github/ad-auction-sim')
from experiments.config import ExperimentConfig
from sim import AdSpot, Bidder

# Parameters
methods = ["gsp"]  # choose from "first_price", "gsp"
n_impressions = 2000
seed = 97
genders = ["male", "female"]

bidder_targeting = {
    "Makeup": {"female": 10.0, "male": 2.0},
    "STEM":   {"female": 1.0,  "male": 1.0},
}


# You can replace custom_valuation's body with your own per-click valuation function
def custom_valuation(
    bidder: Bidder,
    adspot: AdSpot,
    ctrs: list[float],
) -> float:
    return bidder.targeting.get(adspot.tags[0], 0.0)


config = ExperimentConfig(
    n_impressions=n_impressions,
    seed=seed,
    methods=methods,
    genders=genders,
    bidder_targeting=bidder_targeting,
    valuation_fn=custom_valuation,
)
```
